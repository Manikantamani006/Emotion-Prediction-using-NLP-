```python
emotion_no = {}
i=0
for emo in unique_emo:
  emotion_no[emo] = i
  i+=1
emotion_no
```

```python
df['emotion'] = df['emotion'].map(emotion_no)
df
```

```python
df['text'] = df['text'].apply(lambda x:x.lower())
df
```

```python
import string

def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

df['text'] = df['text'].apply(remove_punctuation)
df
```

```python
def remove_numbers(text):
    return ''.join([i for i in text if not i.isdigit()])

df['text'] = df['text'].apply(remove_numbers)
df
```

```python
import re

def remove_urls(text):
    return re.sub(r'https?:\S+|www\.\S+', '', text)

df['text'] = df['text'].apply(remove_urls)
df
```

```python
def remove_emojis_and_special(text):
    # Remove emojis and non-ASCII characters
    text = text.encode('ascii', 'ignore').decode('ascii')
    # Keep only alphanumeric characters and spaces
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    # Replace multiple spaces with a single space and strip
    return ' '.join(text.split())

df['text'] = df['text'].apply(remove_emojis_and_special)
df
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkit')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(text):
  words = word_tokenize(text)
  filtered_words = [word for word in words if word.lower() not in stop_words]
  return ' '.join(filtered_words)
```

```python
import nltk
nltk.download('punkt_tab')

# Re-apply the stopword removal function
df['text'] = df['text'].apply(remove)
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
bow_vectorizer = CountVectorizer()
tfidf_vectorizer = TfidfVectorizer()
```

```python
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB #specifically for word classification involt=ving frequency
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)
```

```python
nb_bow_pred = nb_model.predict(X_test_bow)
nb_bow_accuracy = accuracy_score(y_test, nb_bow_pred)
```

```python
print("Naive Bayes with Bag-of-Words Accuracy:", nb_bow_accuracy)
```

```python
print(classification_report(y_test, nb_bow_pred))
```

```python
nb_model_tfidf = MultinomialNB()
nb_model_tfidf.fit(X_train_tfidf, y_train)
```

```python
pred_tfidf = nb_model_tfidf.predict(X_test_tfidf)
accuracy_tfidf = accuracy_score(y_test, pred_tfidf)
```

```python
print("Naive Bayes with TF-IDF Accuracy:", accuracy_tfidf)
print(classification_report(y_test, pred_tfidf))
```

```python
# Initialize Logistic Regression with the best parameters
best_lr_model = LogisticRegression(C=10.0, max_iter=1000, random_state=42)

# Fit the model on the TF-IDF training data
best_lr_model.fit(X_train_tfidf, y_train)

# Predict on the TF-IDF test data
y_pred_lr = best_lr_model.predict(X_test_tfidf)

# Evaluate accuracy and generate classification report
lr_accuracy = accuracy_score(y_test, y_pred_lr)
print(f"Optimized Logistic Regression Accuracy: {lr_accuracy:.4f}\n")
print("Classification Report:")
print(classification_report(y_test, y_pred_lr, target_names=list(emotion_no.keys())))
```

```python
%%writefile app.py
import streamlit as st
import pickle
import re
import string

# Set Page configuration
st.set_page_config(page_title="Emotion Classifier", page_icon="🎭", layout="centered")

# Load Pickle Artifacts
@st.cache_resource
def load_artifacts():
    with open('best_lr_model.pkl', 'rb') as model_file:
        model = pickle.load(model_file)
    with open('tfidf_vectorizer.pkl', 'rb') as vec_file:
        vectorizer = pickle.load(vec_file)
    with open('emotion_mapping.pkl', 'rb') as mapping_file:
        mapping = pickle.load(mapping_file)

    # Reverse mapping to get emotion name from predicted label index
    reverse_mapping = {v: k for k, v in mapping.items()}
    return model, vectorizer, reverse_mapping

try:
    model, vectorizer, reverse_mapping = load_artifacts()
except Exception as e:
    st.error(f"Error loading artifacts: {e}")

# Preprocessing Functions
def preprocess_text(text):
    # 1. Lowercase
    text = text.lower()
    # 2. Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # 3. Remove numbers
    text = ''.join([i for i in text if not i.isdigit()])
    # 4. Remove URLs
    text = re.sub(r'https?:\S+|www\.\S+', '', text)
    # 5. Remove emojis and non-ASCII characters
    text = text.encode('ascii', 'ignore').decode('ascii')
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    text = ' '.join(text.split())
    return text

# Design the UI/UX
st.title("🎭 Emotion Classifier App")
st.write("Enter a text passage below to predict the underlying emotion.")

user_input = st.text_area("Your Text:", height=150, placeholder="Type how you feel today...")

# Emotion representation mapping
emotion_emojis = {
    'sadness': "😢 Sadness",
    'anger': "😡 Anger",
    'love': "❤️ Love",
    'surprise': "😲 Surprise",
    'fear': "😨 Fear",
    'joy': "😊 Joy"
}

if st.button("Classify Emotion", use_container_width=True):
    if user_input.strip() == "":
        st.warning("Please enter some text before processing.")
    else:
        # Preprocess input text
        clean_text = preprocess_text(user_input)

        # Vectorize text
        vectorized_text = vectorizer.transform([clean_text])

        # Predict class & probabilities
        prediction = model.predict(vectorized_text)[0]
        probabilities = model.predict_proba(vectorized_text)[0]

        predicted_emotion = reverse_mapping[prediction]
        emoji_name = emotion_emojis.get(predicted_emotion, predicted_emotion.capitalize())

        # Display results
        st.success(f"Predicted Emotion: **{emoji_name}**")

        # Display class probabilities visualizer
        st.write("### Confidence Levels:")
        for idx, prob in enumerate(probabilities):
            class_name = reverse_mapping[idx]
            emoji_label = emotion_emojis.get(class_name, class_name.capitalize())
            st.write(f"{emoji_label}: {prob:.2%}")
            st.progress(float(prob))
```
